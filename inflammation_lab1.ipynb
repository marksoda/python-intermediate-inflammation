## Steps
### 1. Select and load the small example

```python
filename = "data/small-01.csv"
inflammation_data = models.load_csv(filename)
print("File:", filename)
print("Patients x days:", inflammation_data.shape)
print(inflammation_data)
```

### 2. Compute the original daily summary statistics

```python
view_data = {
    "average": models.daily_mean(inflammation_data),
    "max": models.daily_max(inflammation_data),
    "min": models.daily_min(inflammation_data),
}
for name, values in view_data.items():
    print(name, values)
```

### 3. Display the original visualisation
Horizontal position represents the day index (starting at 0). The panels show daily average, maximum and minimum inflammation.

```python
views.visualize(view_data)
```

### 4. Change to an inflammation dataset and rerun
This second example makes the changed data path explicit while using the same method.

```python
filename = "data/inflammation-01.csv"
inflammation_data = models.load_csv(filename)
print("File:", filename)
print("Patients x days:", inflammation_data.shape)
view_data = {
    "average": models.daily_mean(inflammation_data),
    "max": models.daily_max(inflammation_data),
    "min": models.daily_min(inflammation_data),
}
views.visualize(view_data)
```

## Checks
Check the numerical summaries against direct NumPy calculations.

```python
assert np.allclose(view_data["average"], np.mean(inflammation_data, axis=0))
assert np.array_equal(view_data["max"], np.max(inflammation_data, axis=0))
assert np.array_equal(view_data["min"], np.min(inflammation_data, axis=0))
print("Summary checks passed.")
```
